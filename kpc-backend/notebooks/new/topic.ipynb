### get_clusters() - list_clusters

```python
from keyphrase_curation.controller.user_attribution \
    import UserAttributionController
from keyphrase_curation.controller.annotation import \
    AnnotationController
import json
from keyphrase_curation.model.cluster import ClusterSorting
from fastapi import HTTPException

username = "daired"
topics = UserAttributionController().get_topics(username)
print("Topics:")
print(json.dumps(topics, indent=4, ensure_ascii=False))

try:
    topic = topics[0]
    
    ac = AnnotationController(topic, username)
    clusters, clusters_meta_info = ac.get_clusters()
    cluster_selection = ac.get_cluster_selection()
    keyphrases_selection = ac.get_keyphrases_selection()
    keyphrases_aliases = ac.get_keyphrases_aliases()
    
    # Reorganizar clusters no formato desejado
    if isinstance(clusters, dict):
        clusters_array = []
        for cluster_id, cluster_data in clusters.items():
            # Pegar os valores correspondentes ao cluster_id
            cluster_sel = cluster_selection.get(str(cluster_id), None)
            
            # keyphrases_selection[cluster_id] é um objeto/dict
            keyphrases_sel_obj = keyphrases_selection.get(str(cluster_id), None)
            if isinstance(keyphrases_sel_obj, dict):
                keyphrases_sel = [
                    keyphrases_sel_obj.get("selected1", ""),
                    keyphrases_sel_obj.get("selected2", ""),
                    keyphrases_sel_obj.get("keyphrase1", ""),
                    keyphrases_sel_obj.get("keyphrase2", "")
                ]
            else:
                keyphrases_sel = []
            
            keyphrases_ali = keyphrases_aliases.get(str(cluster_id), None)
            
            cluster_obj = {
                "cluster_id": cluster_id,
                "cluster_data": cluster_data,
                "cluster_selection": cluster_sel,
                "keyphrases_selection": keyphrases_sel,
                "keyphrases_aliases": keyphrases_ali
            }
            clusters_array.append(cluster_obj)
        
        clusters = clusters_array
    
except HTTPException as http_exc:
    raise http_exc
except Exception as e:
    raise HTTPException(status_code=500, detail=str(e))

print(json.dumps({
    "clusters": clusters,
    "clusters_meta_info": clusters_meta_info
},  indent=4, ensure_ascii=False))
```

## Lidar com Mudanças de Clusters - handle_clustering_change

### move_to_cluster() - move_to_cluster

```python
from keyphrase_curation.controller.user_attribution \
    import UserAttributionController
from keyphrase_curation.controller.annotation import \
    AnnotationController
from keyphrase_curation.model.annotation import \
    AnnotationTask
import json

username = "daired"
topics = UserAttributionController().get_topics(username)
print("Topics:")
print(json.dumps(topics, indent=4, ensure_ascii=False))

ac = AnnotationController(topics[0], username)

keyphrase_id = 3
cluster_id = 11

print("Keyphrase Clustering before moving:")
print(json.dumps(ac.get_clusters(), indent=4, ensure_ascii=False))

print(f"keyphrase_id: {keyphrase_id}")
print(f"cluster_id: {cluster_id}")

try:
    ac.move_to_cluster(int(keyphrase_id), int(cluster_id))
    ac.save(task=AnnotationTask.KEYPHRASE_CLUSTERING)
    print("Keyphrase moved successfully.")

except Exception as e:
    print(f"Error moving keyphrase to cluster: {e}")

print("Keyphrase Clustering after moving:")
print(json.dumps(ac.get_clusters(), indent=4, ensure_ascii=False))
```
