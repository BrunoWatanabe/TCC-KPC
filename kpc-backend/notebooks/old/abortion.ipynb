```python
# spacy
import spacy
from collections import Counter
from string import punctuation
nlp = spacy.load("en_core_web_sm")
def get_hotwords(text):
    result = []
    pos_tag = ['NOUN'] 
    doc = nlp(text.lower()) 
    for token in doc:
        if(token.text in nlp.Defaults.stop_words or token.text in punctuation):
            continue
        if(token.pos_ in pos_tag):
            result.append(token.text)
    return result
new_text = """
When it comes to evaluating the performance of keyword extractors, you can use some of the standard metrics in machine learning: accuracy, precision, recall, and F1 score. However, these metrics don’t reflect partial matches. they only consider the perfect match between an extracted segment and the correct prediction for that tag.
Fortunately, there are some other metrics capable of capturing partial matches. An example of this is ROUGE.
"""
output = set(get_hotwords(all_texts))
most_common_list = Counter(output).most_common(20)
for item in most_common_list:
  print(item[0])
```

Quais são os requisitos da escolha de palavras-chave para classificação de tópicos de argumentação?
- 1 - Unicidade e Diversidade: o conjunto de palavras-chave devem ser semanticamente diversas e diferentes uma da outra, evitando superposição de significado com equilíbrio da cobertura de significados.   - Se uma palavra-chave for semelhante a outra, então somente uma deve ser escolhida.  Por exemplo: conflito armado e guerra são semelhantes, então somente uma deve ser escolhida.

- 2 - Especificidade e generalidade: as palavras-chave devem ser específicas o suficiente para capturar o significado do conteúdo do documento. Também devem ser genéricas o suficientes para serem usadas em diversos documentos.

- 3 - Relevância: as palavras-chave devem ser relevantes para os tópicos ou assuntos que os documentos abordam.

- 4 - Limitação: o conjunto de palavras-chave não devem ser muito extenso para evitar confusão e garantir eficiência. 

- 5 - Consistência: mantenha a consistência nas palavras-chave usadas para classificar os documentos para garantir uma classificação coerente e precisa.

5 - Repetição: use palavras-chave que aparecem com frequência em documentos relevantes.

```python
from __future__ import print_function

import os.path

from google.auth.transport.requests import Request
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build
from googleapiclient.errors import HttpError
import gspread

# If modifying these scopes, delete the file token.json.
SCOPES = ['https://www.googleapis.com/auth/spreadsheets.readonly']

# The ID and range of a sample spreadsheet.
SAMPLE_SPREADSHEET_ID = '1p3dOp-iDKs9mjfzEOmbPcx8OSuMaXJ8r_TxdTR09-YY'
SAMPLE_RANGE_NAME = 'all!A1:E140'


def main():
    """Shows basic usage of the Sheets API.
    Prints values from a sample spreadsheet.
    """
    creds = None
    # The file token.json stores the user's access and refresh tokens, and is
    # created automatically when the authorization flow completes for the first
    # time.
    if os.path.exists('token.json'):
        creds = Credentials.from_authorized_user_file('token.json', SCOPES)
    else:
        creds = Credentials.from_authorized_user_file('/Users/akira/dados/dev/api-keys/client_secret_932013890544-cd4aj9qu8u4p98ksgnock3tc2u3iu3on.apps.googleusercontent.com.json', SCOPES)
    # If there are no (valid) credentials available, let the user log in.
    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
        else:
            flow = InstalledAppFlow.from_client_secrets_file(
                '/Users/akira/dados/dev/api-keys/client_secret_932013890544-cd4aj9qu8u4p98ksgnock3tc2u3iu3on.apps.googleusercontent.com.json', SCOPES)
            creds = flow.run_local_server(port=0)
        # Save the credentials for the next run
        with open('token.json', 'w') as token:
            token.write(creds.to_json())
    try:

        service = build('sheets', 'v4', credentials=creds)

        # Call the Sheets API
        sheet = service.spreadsheets()
        result = sheet.values().get(spreadsheetId=SAMPLE_SPREADSHEET_ID,
                                    range=SAMPLE_RANGE_NAME).execute()
        values = result.get('values', [])
        if not values:
            print('No data found.')
            return
        cluster_set = {}
        cluster_dict = {}
        id = 1
        for row in values:
            if len(row) == 5:
                cluster_set[row[0]] = row[4]
        for cluster in cluster_set.keys():
            x = cluster_set[cluster].split(',')
            x.append(cluster)
            cluster_dict[id] = x
            id = id + 1
        cluster_inverse = {}
        for id in cluster_dict.keys():
            for element_id in cluster_dict[id]:
                cluster_inverse[int(element_id)] = id
        for i in range(1,141):
            if i in cluster_inverse:
                print(f"{values[i-1][2]}\t{cluster_inverse[i]}")
            else:
                print(f"{values[i-1][2]}\t0")
        
    except HttpError as err:
        print(err)


if __name__ == '__main__':
    main()
```
