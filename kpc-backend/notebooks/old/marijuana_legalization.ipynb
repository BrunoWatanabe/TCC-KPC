```python
# spacy
import spacy
from collections import Counter
from string import punctuation
nlp = spacy.load("en_core_web_sm")
def get_hotwords(text):
    result = []
    pos_tag = ['NOUN'] 
    doc = nlp(text.lower()) 
    for token in doc:
        if(token.text in nlp.Defaults.stop_words or token.text in punctuation):
            continue
        if(token.pos_ in pos_tag):
            result.append(token.text)
    return result
new_text = """
When it comes to evaluating the performance of keyword extractors, you can use some of the standard metrics in machine learning: accuracy, precision, recall, and F1 score. However, these metrics don’t reflect partial matches. they only consider the perfect match between an extracted segment and the correct prediction for that tag.
Fortunately, there are some other metrics capable of capturing partial matches. An example of this is ROUGE.
"""
output = set(get_hotwords(all_texts))
most_common_list = Counter(output).most_common(20)
for item in most_common_list:
  print(item[0])
```

Quais são os requisitos da escolha de palavras-chave para classificação de tópicos de argumentação?
- 1 - Unicidade e Diversidade: o conjunto de palavras-chave devem ser semanticamente diversas e diferentes uma da outra, evitando superposição de significado com equilíbrio da cobertura de significados.   - Se uma palavra-chave for semelhante a outra, então somente uma deve ser escolhida.  Por exemplo: conflito armado e guerra são semelhantes, então somente uma deve ser escolhida.

- 2 - Especificidade e generalidade: as palavras-chave devem ser específicas o suficiente para capturar o significado do conteúdo do documento. Também devem ser genéricas o suficientes para serem usadas em diversos documentos.

- 3 - Relevância: as palavras-chave devem ser relevantes para os tópicos ou assuntos que os documentos abordam.

- 4 - Limitação: o conjunto de palavras-chave não devem ser muito extenso para evitar confusão e garantir eficiência. 

- 5 - Consistência: mantenha a consistência nas palavras-chave usadas para classificar os documentos para garantir uma classificação coerente e precisa.

5 - Repetição: use palavras-chave que aparecem com frequência em documentos relevantes.
