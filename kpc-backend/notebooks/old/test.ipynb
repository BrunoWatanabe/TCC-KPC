```python
def make_data(coder_id, labels):
    data = []
    if len(labels) == 0:
        print("Empty keys")
    for id in range(len(labels)):
        data.append((f'Coder_{coder_id}', id, frozenset([labels[id]])))
    return data

# krippendorff's alpha
from nltk.metrics.agreement import AnnotationTask
from nltk.metrics.distance import jaccard_distance

gold_data = make_data(1, gold)
chat_gpt_data = make_data(2, chat_gpt)
chat_gpt_1_2kp_data = make_data(3, chat_gpt_1_2kp)

task = AnnotationTask(data=gold_data + chat_gpt_data, distance=jaccard_distance)
print("Alpha Jaccard gold-chatgpt", round(task.alpha(),2))
task = AnnotationTask(data=gold_data + chat_gpt_1_2kp_data, distance=jaccard_distance)
print("Alpha Jaccard gold-chatgpt_1_2kp", task.alpha())
task = AnnotationTask(data=chat_gpt_data + chat_gpt_1_2kp_data, distance=jaccard_distance)
print("Alpha Jaccard chatgpt-chatgpt_1_2kp", task.alpha())
```
